### 1.1 Identity and Purpose

**Dataset name:** UrbanSound8K  
**Dataset version:** 1.0 (original release)  
**Original creators:** Justin Salamon, Christopher Jacoby, and Juan Pablo Bello  
**Dataset source:** [UrbanSound8K on Kaggle](https://www.kaggle.com/datasets/chrisfilo/urbansound8k)  
**Date retrieved:** August 25, 2026  
**Unit of observation:** One short urban sound recording.

UrbanSound8K is a publicly available dataset containing 8,732 audio recordings of urban environmental sounds. Each recording lasts a maximum of approximately four seconds and belongs to one of ten sound categories.

The ten sound classes are air conditioner, car horn, children playing, dog bark, drilling, engine idling, gun shot, jackhammer, siren, and street music.

The dataset was originally developed to support research in environmental sound classification.

In our IDIG4120 project, we use UrbanSound8K to investigate digital audio representation, transformations, compression, and reconstruction quality.

The dataset is suitable for educational signal-processing experiments. However, our project does not establish its reliability for safety-critical sound detection or commercial applications.
